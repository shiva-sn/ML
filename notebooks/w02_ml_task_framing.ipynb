# W02 – ML Task Framing

## 1. My Lane as an ML Task (Type)

**Lane:** Lane 3 – Structured Content Archetype Clustering

**ML Task Type:** Clustering (Unsupervised Learning)

The objective is to group content pages with similar performance characteristics into meaningful clusters. Unlike classification, clustering does not require predefined labels. It discovers natural groups using metrics such as impressions, sessions, CTR, engagement rate, average position, content age, and word count.

## 2. Target or Proxy

There is **no target label** because clustering is an **unsupervised learning** task.

The model groups pages using features such as impressions_90d, sessions_90d, ctr, avg_position, engagement_rate, scroll_rate, content_age_days, word_count, ai_sessions_90d, content_type, and freshness.

The resulting clusters may later be interpreted as Champions, Rising Stars, Hidden Gems, Stale Visible Pages, Low Engagement Pages, and Weak Demand Pages.

## 3. Success Metric

The clustering quality will be evaluated using:

- Silhouette Score
- Davies–Bouldin Index
- Cluster Visualization
- Cluster Profiling and Interpretation

A successful model should produce well-separated and meaningful clusters.

## 5. Why ML Beats a Fixed Rule Here

Fixed rules only capture predefined conditions, while clustering discovers hidden patterns among pages by considering many features together. This allows similar content pages to be grouped automatically and supports better content review decisions.
